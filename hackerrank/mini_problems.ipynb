```python
#1 

if __name__ == '__main__':
    lst = []
    for _ in range(int(input())):
        name = input()
        score = float(input())
        lst.append([name, score])

    sc = sorted(set([i[1] for i in lst]), reverse=True)


    runner = sc[-2]

    for i in sorted(lst):
        if i[1] == runner:
            print(i[0])
```

```python
#2
if __name__ == '__main__':
    # static inputs
    student_marks = {
        'Om': [100.0, 105.0, 108.0],
        'Sans': [110.0, 108.0, 106.0],
        '101': [100.0, 110.0, 113.0],
    }
    query_name="Sans"
    # use existing query_name
    avg = sum(student_marks[query_name]) / len(student_marks[query_name])
    print(f"{avg:.2f}")
```
